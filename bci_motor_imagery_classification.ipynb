### 1.2 What is a Brain-Computer Interface?

A **Brain-Computer Interface (BCI)** is a system that can translate patterns of brain activity into commands for an external computer or device.

A simplified BCI pipeline can be represented as:

**Brain activity → Signal recording → Signal processing → Machine learning → Device control**

For example, a person could imagine moving their left hand. The brain activity associated with this imagined movement can be recorded using EEG. After processing the signal, a machine learning model can try to determine which movement the person imagined.

In a future assistive technology application, the model's prediction could potentially be converted into a command for a robotic or prosthetic hand.

---

### 1.3 What is EEG?

**Electroencephalography (EEG)** is a non-invasive method for recording electrical activity produced by the brain.

EEG uses electrodes placed on the scalp to measure very small changes in electrical potential over time. Because the brain is continuously active, an EEG recording consists of signals that change over time across multiple electrodes.

An EEG recording can therefore be thought of as a collection of time-dependent signals:

* Multiple **channels** represent different electrode locations.
* Each channel contains measurements collected over **time**.
* The number of measurements depends on the **sampling frequency**.

For example, if an EEG system records at 250 Hz, each channel contains 250 measurements every second.

EEG is particularly useful for BCI research because it can record brain activity without requiring surgery.

---

### 1.4 What is Motor Imagery?

**Motor imagery** is the mental simulation of a movement without physically performing that movement.

For example, a participant may be asked to imagine moving their left hand while keeping their hand completely still. Although there is no actual movement, imagining the movement can produce measurable changes in brain activity.

These changes can be detected in EEG recordings and may contain information about the imagined movement.

In this project, I will focus on distinguishing between **left-hand and right-hand motor imagery**.

### 3.7 Converting Annotations into Events

I have now identified what the event codes in the recording mean.

The annotations tell me when an event happened, while the event code tells me what happened at that time. For example, event `769` represents the left-hand motor imagery cue and event `770` represents the right-hand motor imagery cue.

MNE provides a function that converts these annotations into a numerical event array. This format is useful because I will later use the event positions to extract individual trials from the continuous EEG recording.

In other words, I am not changing the EEG signal itself here. I am only changing the way the event information is organized so that MNE can use it for the next step.

**Source:** The event definitions are taken from the official BCI Competition IV Dataset 2a documentation:
https://www.bbci.de/competition/iv/desc_2a.pdf


````python
# Convert the annotations stored in the raw EEG recording
# into an MNE events array.
#
# This does not change the EEG signal.
# It only converts the event information into a format
# that MNE can use for further processing.

events, event_id = mne.events_from_annotations(
    raw,
    verbose=False
)

# Display the event codes and their numerical IDs.
print("Event ID mapping:")
print(event_id)

# Display the shape of the events array.
# The first number tells us how many events there are.
# The second number tells us how much information is stored
# for each event.
print("\nEvents array shape:")
print(events.shape)

# Display the first 10 events so that we can see
# how the information is represented.
print("\nFirst 10 events:")
print(events[:10])
````

#### Understanding the MNE Event IDs

One important detail can be seen in the output above.

The original dataset uses event codes such as `769`, `770`, `771`, and `772`. However, MNE assigns its own internal numerical IDs to these events.

For example:

| Original event code | Meaning    | MNE internal ID |
| ------------------- | ---------- | --------------: |
| `769`               | Left hand  |             `7` |
| `770`               | Right hand |             `8` |
| `771`               | Foot       |             `9` |
| `772`               | Tongue     |            `10` |

This does **not** mean that MNE changed the original event codes.

Instead, MNE creates its own numerical identifiers so that it can work with the events internally. The `event_id` dictionary shows the relationship between the original event codes and MNE's internal IDs.

For example:

```python
event_id['769']
```

returns:

```text
7
```

Therefore, when I see `7` in the third column of the `events` array, I know that it represents the original event code `769`, which corresponds to left-hand motor imagery.

This distinction is important because I will use MNE's internal event IDs when creating epochs in the next step.


### 3.8 Selecting Left- and Right-Hand Trials

The dataset contains four different motor imagery classes:

* Left hand
* Right hand

* Tongue

For this project, I am focusing only on **left-hand and right-hand motor imagery**.

From the previous section, I know that MNE assigned the following internal IDs:

I will therefore select only these two types of events.

````python
# Get the MNE internal IDs corresponding to the
# left-hand and right-hand motor imagery events.
left_hand_id = event_id['769']
right_hand_id = event_id['770']

# Select only the events belonging to the two classes
# that we want to use in this binary classification project.
left_hand_events = events[events[:, 2] == left_hand_id]
right_hand_events = events[events[:, 2] == right_hand_id]

# Display the number of events in each class.
print("Number of left-hand events:", len(left_hand_events))
print("Number of right-hand events:", len(right_hand_events))
````

### 3.9 Creating EEG Epochs

The EEG recording is stored as one long, continuous signal. However, for classification, I need to work with individual trials.

An **epoch** is a short segment of EEG extracted from the continuous recording around a specific event.

For this project, I will extract **4 seconds of EEG data after each motor imagery cue**.

According to the official experiment protocol, the cue appears 2 seconds after the beginning of a trial, and the participant continues the motor imagery until approximately 6 seconds. Therefore, using the cue onset as the starting point gives me a 4-second period corresponding to the motor imagery task.

I will create epochs for both:

* Left-hand motor imagery
* Right-hand motor imagery

I will also exclude the three EOG channels because they are intended for detecting eye-related artifacts and should not be used as input features for the classification model.

The resulting epochs will contain individual EEG trials that can later be filtered, analyzed, and converted into features for machine learning.

**Source:** The timing and channel information are based on the official BCI Competition IV Dataset 2a documentation:
https://www.bbci.de/competition/iv/desc_2a.pdf


````python
# Define the event IDs that we want to use for epoching.
# We already identified these IDs in the previous section.
motor_event_id = {
    'left_hand': left_hand_id,
    'right_hand': right_hand_id
}

# Create epochs from the continuous EEG recording.
#
# tmin=0 means that the epoch starts exactly when
# the motor imagery cue appears.
#
# tmax=4 means that we keep approximately 4 seconds
# of EEG data after the cue.
#
# baseline=None means that we are not applying
# baseline correction at this stage.
epochs = mne.Epochs(
    raw,
    events,
    event_id=motor_event_id,
    tmin=0,
    tmax=4,
    baseline=None,
    preload=True,
    verbose=False
)

# Remove the three EOG channels because they should not
# be used as input signals for the classification task.
epochs.drop_channels(['EOG-left', 'EOG-central', 'EOG-right'])

# Display information about the resulting epochs.
print(epochs)
````

### 3.12 Filtering the EEG Signal

The raw EEG signal contains many different frequency components.

Not all of these components are equally useful for distinguishing between left-hand and right-hand motor imagery. The signal may also contain slow changes, high-frequency noise, and other unwanted components.

For this reason, I will apply a **band-pass filter** to keep only the frequency range that is relevant for the analysis.

The original recording was acquired with a frequency range of approximately **0.5–100 Hz**. For this motor imagery classification task, I will focus on the **8–30 Hz** range.

This range is particularly useful because it contains the **mu rhythm (approximately 8–13 Hz)** and **beta rhythm (approximately 13–30 Hz)**. These frequency bands are commonly associated with sensorimotor activity and are therefore relevant to motor imagery analysis.

The goal of filtering is not to create new information. Instead, I am removing frequency components that are less useful for the specific question I am trying to answer.

In simple terms:

```text
Original EEG
     ↓
Keep frequencies between 8–30 Hz
     ↓
Filtered EEG
     ↓
Focus on sensorimotor-related activity
```

I will apply the filter to a copy of the epochs so that I can keep the original epoch data unchanged.

**Note:** This is a practical learning/demo choice for this project. Different EEG studies may use different frequency ranges depending on their research question.


````python
# Create a copy of the epochs so that the original data
# remains unchanged.
filtered_epochs = epochs.copy()

# Apply a band-pass filter between 8 and 30 Hz.
#
# Frequencies below 8 Hz and above 30 Hz are attenuated,
# while the frequencies inside this range are kept.
filtered_epochs.filter(
    l_freq=8,
    h_freq=30,
    verbose=False
)

print(filtered_epochs)
````

````python
# Get the numerical EEG data from the filtered epochs.
filtered_data = filtered_epochs.get_data()

# Select the first channel from the first trial.
filtered_signal = filtered_data[0, 0]

# Create the time axis using the sampling frequency.
time = np.arange(len(filtered_signal)) / sfreq

# Plot the filtered EEG signal.
plt.figure(figsize=(12, 4))

plt.plot(time, filtered_signal)

plt.xlabel("Time (seconds)")
plt.ylabel("EEG amplitude (V)")
plt.title("Filtered EEG Signal - First Channel, First Trial")

plt.show()
````

### 3.13 Frequency Analysis with Power Spectral Density

After filtering the EEG signal to the 8–30 Hz range, I now want to examine how the signal's power is distributed across different frequencies.

Looking at the EEG signal only in the time domain tells me how the signal changes over time, but it does not directly tell me which frequencies are dominant.

To investigate this, I will use **Power Spectral Density (PSD)**.

PSD describes how much power a signal has at different frequencies. In other words, it allows me to see which frequency components contribute more strongly to the EEG signal.

This is particularly useful for motor imagery because the sensorimotor system is associated with activity in frequency bands such as:

- **Mu rhythm:** approximately 8–13 Hz
- **Beta rhythm:** approximately 13–30 Hz

I will calculate the PSD of the first EEG channel from the first trial and visualize it.

The x-axis will represent frequency in Hertz (Hz), while the y-axis will represent the power of the signal at each frequency.

This will allow me to move from simply looking at the EEG waveform to examining its frequency content.

````python
# Calculate the Power Spectral Density (PSD)
# of the first trial using MNE.
spectrum = filtered_epochs.compute_psd(
    method="welch",
    fmin=8,
    fmax=30,
    verbose=False
)

# Get the numerical PSD values.
psd = spectrum.get_data()

# Get the frequency values corresponding to the PSD.
frequencies = spectrum.freqs

# Select the first trial and the first EEG channel.
first_channel_psd = psd[0, 0]

# Plot the Power Spectral Density.
plt.figure(figsize=(12, 4))

plt.plot(frequencies, first_channel_psd)

plt.xlabel("Frequency (Hz)")
plt.ylabel("Power")
plt.title("Power Spectral Density - First Channel, First Trial")

plt.show()
````

### 3.14 Comparing Left- and Right-Hand Frequency Power

So far, I have examined the frequency content of a single EEG trial.

However, a single trial is not enough to determine whether the EEG contains information related to the imagined movement.

I now want to compare the frequency content of the **left-hand** and **right-hand** motor imagery trials.

For this comparison, I will calculate the Power Spectral Density (PSD) for all selected trials and then calculate the average PSD separately for each class.

This will allow me to ask a more meaningful question:

> Are there differences in the frequency power of left-hand and right-hand motor imagery trials?

I will start with the first EEG channel so that the comparison remains easy to interpret.

The resulting plot will show:

- the average PSD for left-hand trials
- the average PSD for right-hand trials

If the two curves show differences at certain frequencies, this may indicate that those frequency ranges contain information that could be useful for distinguishing the two motor imagery classes.

This does not mean that the two classes can already be classified successfully. It is an exploratory step that helps me understand the structure of the EEG data before building a machine learning model.

````python
# Calculate the PSD for all trials in the filtered EEG data.
spectrum = filtered_epochs.compute_psd(
    method="welch",
    fmin=8,
    fmax=30,
    verbose=False
)

# Get the numerical PSD values.
psd = spectrum.get_data()

# Get the corresponding frequency values.
frequencies = spectrum.freqs

# Create a boolean mask for left-hand trials.
left_mask = epochs.events[:, 2] == left_hand_id

# Create a boolean mask for right-hand trials.
right_mask = epochs.events[:, 2] == right_hand_id

# Calculate the average PSD of the first channel
# across all left-hand trials.
left_psd_mean = psd[left_mask, 0, :].mean(axis=0)

# Calculate the average PSD of the first channel
# across all right-hand trials.
right_psd_mean = psd[right_mask, 0, :].mean(axis=0)

# Plot the average PSD of both classes.
plt.figure(figsize=(12, 4))

plt.plot(
    frequencies,
    left_psd_mean,
    label="Left hand"
)

plt.plot(
    frequencies,
    right_psd_mean,
    label="Right hand"
)

plt.xlabel("Frequency (Hz)")
plt.ylabel("Power")
plt.title("Average PSD Comparison - First EEG Channel")

plt.legend()

plt.show()
````

### 3.20 Extracting Mu Band Power

The previous analysis showed that some of the largest differences between left-hand and right-hand motor imagery appeared around 11–12 Hz.

This frequency range falls within the **mu rhythm**, which is commonly defined approximately between 8 and 13 Hz.

Instead of only visualizing the PSD, I now want to convert this frequency information into numerical features that can later be used by a machine learning model.

One simple feature is **band power**.

Band power represents the amount of signal power contained within a specific frequency range.

For this project, I will calculate the power in the 8–13 Hz range for each trial and each EEG channel.

The result will provide one numerical value for the mu-band power of each channel in each trial.

The data will therefore change from:

```text
trial × channel × frequency
```

to:

```text
trial × channel
```

This is an important step because machine learning models need numerical features for each trial rather than the original continuous EEG signal.

I will first calculate the mu-band power for all 144 trials and 22 EEG channels.


````python
# Compute the PSD in the mu frequency range (8–13 Hz)
mu_spectrum = filtered_epochs.compute_psd(
    method="welch",
    fmin=8,
    fmax=13,
    verbose=False
)

# Get the PSD values and corresponding frequencies
mu_psd = mu_spectrum.get_data()
mu_frequencies = mu_spectrum.freqs

print("Mu PSD shape:")
print(mu_psd.shape)

print("\nNumber of frequency points:")
print(len(mu_frequencies))
````

````python
print(mu_frequencies)
````

### 3.21 Calculating Mu Band Power

I now have 20 PSD values for each EEG channel within the 8–13 Hz frequency range.

However, I do not want to keep all 20 frequency values as separate features. Instead, I want to summarize the total power contained within the mu frequency band.

To do this, I will calculate the area under the PSD curve across the 8–13 Hz range.

This gives me one **mu-band power** value for each trial and each EEG channel.

Conceptually:

```text id="w7b4yx"
PSD values across 8–13 Hz
            ↓
Area under the PSD curve
            ↓
One mu-band power value
```

I will use numerical integration with the trapezoidal rule.

The resulting feature matrix will have the shape:

```text id="qjy4lq"
(number of trials, number of EEG channels)
```

For this dataset, this means:

```text id="9iyd5v"
144 trials × 22 EEG channels
```

Each value represents the mu-band power of one EEG channel during one trial.


````python
# Calculate the area under the PSD curve
# across the 8–13 Hz mu band.
mu_band_power = np.trapezoid(
    mu_psd,
    x=mu_frequencies,
    axis=2
)

print("Mu band power shape:")
print(mu_band_power.shape)
````

### 3.31 Feature Scaling

Before training a machine learning model, I will standardize the EEG features using `StandardScaler`.

At this point, the feature matrix contains 44 features for each trial:

```text
22 mu-band power features
+
22 beta-band power features
=
44 features
```

These features do not necessarily have the same numerical scale.

For example, one feature might have values around:

```text
7 × 10⁻¹²
```

while another feature might have values around:

```text
3 × 10⁻¹³
```

The difference in numerical scale does not necessarily mean that one feature contains more useful information. It can simply mean that the EEG power values for those features are naturally measured at different scales.

Some machine learning algorithms are affected by the scale of the input features. If one feature has much larger numerical values than another, the model may give it more influence because of its scale rather than because it contains more useful information.

To avoid this, I will standardize the features.

`StandardScaler` uses the following formula:

```text
standardized value = (value - mean) / standard deviation
```

This changes the scale of each feature so that, based on the training data, its values are centered around 0 and have a standard deviation of approximately 1.

For example, a feature might originally look like:

```text
[7.0 × 10⁻¹², 8.0 × 10⁻¹², 6.0 × 10⁻¹², ...]
```

After standardization, it could look more like:

```text
[-0.2, 0.5, -0.8, ...]
```

The important point is that the relative information in the feature is preserved, but the numerical scale is changed.

### Why do we fit the scaler only on the training data?

This is an important part of the machine learning workflow.

We already divided our data into:

```text
Training data → 115 trials
Test data     → 29 trials
```

The training data is the data from which the model is allowed to learn.

The test data should represent **unseen data** that the model has not used during training.

When `StandardScaler` is fitted, it calculates the mean and standard deviation of each feature.

Therefore, I should calculate these values using only `X_train`.

The process is:

```text
X_train
   ↓
fit
   ↓
Learn the mean and standard deviation
   ↓
transform
   ↓
X_train_scaled
```

For the test data, I do not calculate new means and standard deviations.

Instead, I use the values learned from the training data:

```text
X_test
   ↓
transform using the existing scaler
   ↓
X_test_scaled
```

In other words:

```text
TRAINING DATA
"Learn the scale from this data."

TEST DATA
"Use the scale you already learned."
```

If I fitted the scaler using the entire dataset, information from the test set would indirectly enter the preprocessing step. This is called **data leakage**.

The goal is to keep the test data completely separate until the final evaluation.

Therefore, the correct order is:

```text
All data
   ↓
Train / Test split
   ↓
      ┌─────────────────┐
      │                 │
   Training           Test
      │                 │
      ↓                 │
    fit                │
      ↓                 │
   transform            │
      ↓                 ↓
X_train_scaled     transform
                   using the
                   same scaler
                         ↓
                  X_test_scaled
```

This allows us to evaluate the model more honestly on data that was not used to determine the feature scaling.


````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("X_train_scaled shape:", X_train_scaled.shape)
print("X_test_scaled shape:", X_test_scaled.shape)
````

### 4.1 Creating a Feature Extraction Function

The previous analysis was performed manually for the `A01T` recording.

To apply the same preprocessing and feature extraction steps consistently to all nine subjects, I will organize these steps into a reusable function.

The function will perform the same operations used in the previous analysis:

1. Load the GDF recording.
2. Extract the events.
3. Select left- and right-hand motor imagery trials.
4. Create 4-second epochs.
5. Remove the EOG channels.
6. Apply an 8–30 Hz band-pass filter.
7. Extract mu-band power (8–13 Hz).
8. Extract beta-band power (13–30 Hz).
9. Combine the features.
10. Create the corresponding class labels.

At this stage, the preprocessing and feature extraction method is intentionally kept unchanged. This allows us to first investigate the effect of increasing the number of subjects without changing the feature methodology.


````python
def extract_features(file_path):

    # Load the GDF recording
    raw = mne.io.read_raw_gdf(
        file_path,
        preload=True,
        verbose=False
    )

    # Convert annotations into events
    events, event_id = mne.events_from_annotations(
        raw,
        verbose=False
    )

    # Get event IDs for left- and right-hand imagery
    left_hand_id = event_id["769"]
    right_hand_id = event_id["770"]

    # Define the two classes
    motor_event_id = {
        "left_hand": left_hand_id,
        "right_hand": right_hand_id
    }

    # Create epochs from cue onset to 4 seconds
    epochs = mne.Epochs(
        raw,
        events,
        event_id=motor_event_id,
        tmin=0,
        tmax=4,
        baseline=None,
        preload=True,
        verbose=False
    )

    # Remove EOG channels
    epochs.drop_channels([
        "EOG-left",
        "EOG-central",
        "EOG-right"
    ])

    # Apply 8–30 Hz band-pass filter
    filtered_epochs = epochs.copy()

    filtered_epochs.filter(
        l_freq=8,
        h_freq=30,
        verbose=False
    )

    # Calculate mu-band PSD
    mu_spectrum = filtered_epochs.compute_psd(
        method="welch",
        fmin=8,
        fmax=13,
        verbose=False
    )

    mu_psd = mu_spectrum.get_data()
    mu_frequencies = mu_spectrum.freqs

    # Calculate mu-band power
    mu_band_power = np.trapezoid(
        mu_psd,
        x=mu_frequencies,
        axis=2
    )

    # Calculate beta-band PSD
    beta_spectrum = filtered_epochs.compute_psd(
        method="welch",
        fmin=13,
        fmax=30,
        verbose=False
    )

    beta_psd = beta_spectrum.get_data()
    beta_frequencies = beta_spectrum.freqs

    # Calculate beta-band power
    beta_band_power = np.trapezoid(
        beta_psd,
        x=beta_frequencies,
        axis=2
    )

    # Combine mu and beta features
    X = np.concatenate(
        [mu_band_power, beta_band_power],
        axis=1
    )

    # Create class labels
    labels = []

    for event in epochs.events[:, 2]:

        if event == left_hand_id:
            labels.append(0)
        else:
            labels.append(1)

    y = np.array(labels)

    return X, y
````

# 5. Common Spatial Pattern (CSP)

### 5.1 Understanding Spatial Patterns with CSP

So far, I have mainly looked at the frequency power of individual EEG channels.

However, brain activity is not necessarily represented by one electrode working independently. Information about a movement can appear as a pattern across multiple EEG channels.

For example, imagine that we have three EEG channels:

```text
Left-hand imagery:

Channel A → 0.8
Channel B → 0.2
Channel C → 0.7
```

For right-hand imagery, the values might be different:

```text
Right-hand imagery:

Channel A → 0.3
Channel B → 0.9
Channel C → 0.1
```

Instead of looking at each channel separately, we can ask:

> **Can a specific combination of values from different EEG channels help us distinguish left-hand imagery from right-hand imagery?**

This is the idea behind **Common Spatial Pattern (CSP)**.

CSP learns combinations of EEG channels that create spatial patterns that are more different between the two classes.

For example, a spatial pattern could be represented by a combination such as:

```text
0.8 × Channel A
- 0.4 × Channel B
+ 0.7 × Channel C
```

This does **not** mean that these channels physically trigger the movement.

Instead, it means that this combination of EEG signals can form a pattern that helps distinguish the two motor imagery classes.

The goal is therefore to find spatial patterns that are particularly useful for separating:

```text
Left-hand imagery
        vs.
Right-hand imagery
```

After finding these patterns, we can extract numerical features from them and give those features to a machine learning model.

The overall idea is:

```text
EEG signals
    ↓
Multiple EEG channels
    ↓
CSP
    ↓
Discriminative spatial patterns
    ↓
Numerical features
    ↓
Machine learning model
    ↓
Left hand / Right hand
```

This allows the model to use not only the frequency information of individual channels, but also the relationships between different EEG channels.


### 5.2 Preparing EEG Epochs for CSP

The previous feature extraction function converted each EEG trial into 44 numerical features based on mu and beta band power.

For CSP, I need to work with the actual EEG signals instead of only the extracted frequency features.

Therefore, I will create a new function that loads each recording, extracts the left- and right-hand trials, removes the EOG channels, and returns the epoched EEG data together with the class labels.

The data returned by this function will have the following structure:

```text
Number of trials × EEG channels × time samples
```

For each subject, this should give:

```text
144 trials × 22 EEG channels × 1001 samples
```

These epoched EEG signals will later be used to learn spatial patterns with CSP.


````python
def extract_epochs_for_csp(file_path):

    # Load the GDF recording
    raw = mne.io.read_raw_gdf(
        file_path,
        preload=True,
        verbose=False
    )

    # Convert annotations into events
    events, event_id = mne.events_from_annotations(
        raw,
        verbose=False
    )

    # Get event IDs for left- and right-hand imagery
    left_hand_id = event_id["769"]
    right_hand_id = event_id["770"]

    # Define the two classes
    motor_event_id = {
        "left_hand": left_hand_id,
        "right_hand": right_hand_id
    }

    # Create epochs from cue onset to 4 seconds
    epochs = mne.Epochs(
        raw,
        events,
        event_id=motor_event_id,
        tmin=0,
        tmax=4,
        baseline=None,
        preload=True,
        verbose=False
    )

    # Remove EOG channels
    epochs.drop_channels([
        "EOG-left",
        "EOG-central",
        "EOG-right"
    ])

    # Apply the same 8–30 Hz band-pass filter
    epochs.filter(
        l_freq=8,
        h_freq=30,
        verbose=False
    )

    # Get EEG data
    X = epochs.get_data()

    # Create class labels
    labels = []

    for event in epochs.events[:, 2]:

        if event == left_hand_id:
            labels.append(0)
        else:
            labels.append(1)

    y = np.array(labels)

    return X, y
````

### 5.3.1 Understanding CSP Filters and Patterns

CSP learns different ways of combining the 22 EEG channels.

It is useful to think of the EEG channels as different ingredients.

The CSP filter is like a **recipe** that tells us how to combine these channels.

For example:

```text
Channel A → take a little
Channel B → take more
Channel C → reduce
...
```

This is the basic idea behind `csp.filters_`.

The `filters_` values tell CSP **how to combine the EEG channels** to create new spatial components.

The `patterns_` values help us understand **which EEG channels are more related to these spatial components**.

In simple terms:

```text
filters_
    ↓
"How should I combine the channels?"

patterns_
    ↓
"Which channels are more involved in this pattern?"
```

We do not need to interpret every individual number in these arrays.

The important idea is that CSP takes the original EEG signals from 22 channels and creates new components that may make the difference between left-hand and right-hand motor imagery easier to detect.

The overall process is:

```text
22 EEG channels
        ↓
       CSP
        ↓
6 CSP components
        ↓
6 numerical features per trial
        ↓
Machine learning model
        ↓
Left hand / Right hand
```

Therefore, CSP allows the machine learning model to use information from **combinations of EEG channels**, rather than looking at each channel separately.


### 5.4 Inspecting CSP Features

After applying CSP, each EEG trial is represented by 6 new features instead of the original 22 EEG channels.

The original data had:

```text
144 trials × 22 EEG channels × 1001 time samples
```

After applying CSP, the data becomes:

```text
144 trials × 6 CSP features
```

Each trial is now represented by 6 numerical values.

For example, a trial can be represented as:

```text
Trial 1 → [feature 1, feature 2, feature 3, feature 4, feature 5, feature 6]
```

These features summarize information from combinations of the original EEG channels.

The number of features is 6 because I used:

```python
n_components=6
```

when creating the CSP model.

At this stage, I will inspect the shape and some example values of the CSP feature matrix.

The purpose of this step is simply to understand what the data looks like before using it for machine learning.


````python
print("CSP feature matrix shape:")
print(X_A01_CSP_features.shape)

print("\nFirst trial:")
print(X_A01_CSP_features[0])

print("\nFirst 5 trials:")
print(X_A01_CSP_features[:5])
````

### 5.5 Preparing Training and Test Data for CSP

Before training the machine learning model, the EEG trials will be divided into training and test sets.

The important point is that CSP must also be learned only from the training data.

If CSP is learned using the entire dataset before the split, information from the test data can influence the learned spatial patterns.

Therefore, the correct order is:

```text
EEG epochs
    ↓
Train / Test split
    ↓
Training data → CSP.fit()
    ↓
Training data → CSP.transform()
    ↓
Test data → CSP.transform()
    ↓
SVM
    ↓
Prediction
```

This ensures that the test data is not used when learning the CSP spatial patterns.

For this experiment, I will continue using the A01 subject so that the CSP + SVM method can first be tested on the same subject-level data.


````python
from sklearn.model_selection import train_test_split

X_train_CSP, X_test_CSP, y_train_CSP, y_test_CSP = train_test_split(
    X_A01_CSP,
    y_A01_CSP,
    test_size=0.20,
    random_state=42,
    stratify=y_A01_CSP
)

print("Training EEG data shape:")
print(X_train_CSP.shape)

print("\nTest EEG data shape:")
print(X_test_CSP.shape)

print("\nTraining labels shape:")
print(y_train_CSP.shape)

print("\nTest labels shape:")
print(y_test_CSP.shape)
````

#### 5.5.1 Training CSP on the Training Data

The training data contains 115 EEG trials.

I will now train the CSP model using only these training trials.

The CSP model will learn spatial patterns that can help distinguish left-hand imagery from right-hand imagery.

The test data will not be used during this learning step.


````python
from mne.decoding import CSP

csp_model = CSP(
    n_components=6,
    reg=None,
    log=True,
    norm_trace=False
)

csp_model.fit(
    X_train_CSP,
    y_train_CSP
)

print("CSP training completed.")
````

````python
X_train_CSP_features = csp_model.transform(
    X_train_CSP
)

X_test_CSP_features = csp_model.transform(
    X_test_CSP
)

print("Training CSP features shape:")
print(X_train_CSP_features.shape)

print("\nTest CSP features shape:")
print(X_test_CSP_features.shape)
````

### 5.6 Training SVM with CSP Features

The CSP transformation has reduced each EEG trial to 6 numerical features.

I will now use these features to train a Support Vector Machine (SVM) classifier.

The SVM will learn the relationship between the 6 CSP features and the two motor imagery classes:

```text
0 → Left hand
1 → Right hand
```

The model will be trained only on the training data.

After training, it will be used to predict the classes of the unseen test trials.


````python
from sklearn.svm import SVC

csp_svm = SVC(
    kernel="linear",
    random_state=42
)

csp_svm.fit(
    X_train_CSP_features,
    y_train_CSP
)

print("CSP + SVM training completed.")
````

````python
y_pred_CSP = csp_svm.predict(
    X_test_CSP_features
)

print("Predicted labels:")
print(y_pred_CSP)

print("\nNumber of predictions:")
print(len(y_pred_CSP))
````

### 5.7 Evaluating CSP + SVM

The SVM has now predicted the class of all 29 test trials.

However, the predictions alone do not tell us how many predictions were correct.

I will compare the predicted labels with the actual labels from the test set.

The accuracy will show the proportion of test trials that were classified correctly.

For example, an accuracy of 0.70 means that 70% of the test trials were classified correctly.


````python
from sklearn.metrics import accuracy_score

csp_accuracy = accuracy_score(
    y_test_CSP,
    y_pred_CSP
)

print("CSP + SVM Accuracy:", csp_accuracy)
````

### 5.8 Confusion Matrix for CSP + SVM

Accuracy shows the overall proportion of correctly classified trials.

However, it does not show which classes were predicted correctly or incorrectly.

A confusion matrix allows us to examine the predictions for the two classes separately:

```text
0 → Left hand
1 → Right hand
```

The rows represent the **true classes**, while the columns represent the **predicted classes**.

This allows us to see how many left-hand trials were correctly classified, how many were confused with right-hand trials, and vice versa.


````python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm_CSP = confusion_matrix(
    y_test_CSP,
    y_pred_CSP
)

print("Confusion matrix:")
print(cm_CSP)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_CSP,
    display_labels=["Left hand", "Right hand"]
)

disp.plot()

plt.title("Confusion Matrix - CSP + SVM")
plt.show()
````

### Interpretation of the Confusion Matrix

The confusion matrix shows that the CSP + SVM model correctly classified:

* 11 left-hand trials
* 10 right-hand trials

It incorrectly classified:

* 4 left-hand trials as right-hand trials
* 4 right-hand trials as left-hand trials

Therefore, the model correctly classified 21 out of 29 test trials, resulting in an accuracy of approximately **72.4%**.

The number of errors is the same for both directions, with 4 left-hand trials classified as right-hand and 4 right-hand trials classified as left-hand.

These results are based on the A01 subject and a test set of 29 trials, so they should be considered an initial experiment rather than a general performance estimate.


# 6. Applying CSP to All Subjects

### 6.1 Preparing EEG Epochs for All Subjects

The previous CSP experiment used only subject A01.

The dataset contains recordings from nine subjects, so I will now extend the CSP analysis to all subjects.

For each subject, I will use the same preprocessing steps:

* Load the GDF recording
* Extract left- and right-hand motor imagery trials
* Remove the EOG channels
* Apply an 8–30 Hz band-pass filter
* Keep the epoched EEG signals

Each subject contributes 144 trials recorded from 22 EEG channels.

The expected data for each subject is therefore:

```text
144 trials × 22 EEG channels × 1001 time samples
```

After combining all nine subjects, the expected dataset will contain:

```text
1296 trials × 22 EEG channels × 1001 time samples
```

The subject identity will also be stored so that subject-wise evaluation can be performed later.


````python
import os

data_path = (
    "/kaggle/input/datasets/prashastham/"
    "bci-competition-iv-dataset-2a/Train"
)

all_epochs_CSP = []
all_labels_CSP = []
all_subject_ids_CSP = []

for subject_number in range(1, 10):

    subject_name = f"A{subject_number:02d}"

    file_path = os.path.join(
        data_path,
        f"{subject_name}T.gdf"
    )

    print(f"Processing {subject_name}...")

    X_subject, y_subject = extract_epochs_for_csp(
        file_path
    )

    all_epochs_CSP.append(X_subject)
    all_labels_CSP.append(y_subject)

    for i in range(len(y_subject)):
        all_subject_ids_CSP.append(subject_name)

print("CSP data extraction completed.")
````

### 6.2 Combining All Subjects

The EEG epochs have now been extracted from all nine subjects.

Each subject contributes:

* 144 trials
* 22 EEG channels
* 1001 time samples

I will now combine the data from all subjects into a single EEG dataset.

The expected dimensions are:

```text
9 × 144 = 1296 trials

1296 trials × 22 EEG channels × 1001 time samples
```

The class labels and subject identities will be stored separately.


````python
X_all_CSP = np.concatenate(
    all_epochs_CSP,
    axis=0
)

y_all_CSP = np.concatenate(
    all_labels_CSP,
    axis=0
)

subject_ids_CSP = np.array(
    all_subject_ids_CSP
)

print("Combined EEG data shape:")
print(X_all_CSP.shape)

print("\nCombined labels shape:")
print(y_all_CSP.shape)

print("\nSubject IDs shape:")
print(subject_ids_CSP.shape)
````

### 6.3 Train/Test Split for All Subjects

The combined dataset now contains 1296 EEG trials from nine subjects.

I will divide these trials into training and test sets using an 80/20 split.

The class distribution will be preserved using stratification.

At this stage, the split is performed across individual trials. Therefore, recordings from the same subject may appear in both the training and test sets.

This allows us to directly compare the multi-subject result with the previous A01 experiment.

A subject-wise evaluation will be performed later to test generalization to unseen subjects.


````python
from sklearn.model_selection import train_test_split

X_train_all_CSP, X_test_all_CSP, y_train_all_CSP, y_test_all_CSP = train_test_split(
    X_all_CSP,
    y_all_CSP,
    test_size=0.20,
    random_state=42,
    stratify=y_all_CSP
)

print("Training EEG data shape:")
print(X_train_all_CSP.shape)

print("\nTest EEG data shape:")
print(X_test_all_CSP.shape)

print("\nTraining labels shape:")
print(y_train_all_CSP.shape)

print("\nTest labels shape:")
print(y_test_all_CSP.shape)
````

### 6.4 Training CSP with All Subjects

The training set contains 1036 EEG trials from all nine subjects.

I will now train a new CSP model using only the training data.

The CSP model will learn spatial patterns that help distinguish left-hand imagery from right-hand imagery.

The test data will not be used during this learning step.

I will use 6 CSP components, consistent with the previous A01 experiment.


````python
csp_all = CSP(
    n_components=6,
    reg=None,
    log=True,
    norm_trace=False
)

csp_all.fit(
    X_train_all_CSP,
    y_train_all_CSP
)

print("CSP training with all subjects completed.")
````

````python
X_train_all_CSP_features = csp_all.transform(
    X_train_all_CSP
)

X_test_all_CSP_features = csp_all.transform(
    X_test_all_CSP
)

print("Training CSP features shape:")
print(X_train_all_CSP_features.shape)

print("\nTest CSP features shape:")
print(X_test_all_CSP_features.shape)
````

### 6.5 Training SVM with CSP Features

The CSP transformation has converted the EEG data into 6 numerical features for each trial.

I will now train a Linear SVM classifier using these CSP features.

The model will learn from the 1036 training trials and then predict the classes of the 260 unseen test trials.

The two classes are:

```text id="d2f6q8"
0 → Left hand
1 → Right hand
```

The goal is to determine whether the spatial information extracted by CSP can help distinguish between the two motor imagery classes across multiple subjects.


````python
from sklearn.svm import SVC

csp_svm_all = SVC(
    kernel="linear",
    random_state=42
)

csp_svm_all.fit(
    X_train_all_CSP_features,
    y_train_all_CSP
)

print("CSP + SVM training with all subjects completed.")
````

````python
y_pred_CSP_all = csp_svm_all.predict(
    X_test_all_CSP_features
)

print("Number of predictions:")
print(len(y_pred_CSP_all))

print("\nFirst 20 predictions:")
print(y_pred_CSP_all[:20])
````

### 6.6 Evaluating CSP + SVM

The CSP + SVM model has now generated predictions for all 260 test trials.

I will calculate the classification accuracy to measure how many of these trials were classified correctly.

Accuracy is calculated as:

correct predictions / total predictions

````python
from sklearn.metrics import accuracy_score

csp_all_accuracy = accuracy_score(
    y_test_all_CSP,
    y_pred_CSP_all
)

print(f"CSP + SVM Accuracy: {csp_all_accuracy:.4f}")
print(f"CSP + SVM Accuracy: {csp_all_accuracy * 100:.2f}%")
````
